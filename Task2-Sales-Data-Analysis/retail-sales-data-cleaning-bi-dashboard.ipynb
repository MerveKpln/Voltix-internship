****Data Cleaning****

```python
# Converts dates to datetime
df['Invoice Date Key']  = pd.to_datetime(df['Invoice Date Key'], format='%m/%d/%Y')
df['Delivery Date Key'] = pd.to_datetime(df['Delivery Date Key'], format='%m/%d/%Y')

df[['Invoice Date Key', 'Delivery Date Key']].dtypes
```

```python
#Missing delivery date
missing_delivery = df[df['Delivery Date Key'].isna()]
missing_delivery[['Sale Key', 'Invoice Date Key', 'Customer Key', 'WWI Invoice ID','Delivery Date Key']]
```

```python
# 13 sales from the last day in the dataset (2016-05-31) had no delivery date yet —
# most likely because delivery had not occurred by the data extraction date.
print(missing_delivery['Invoice Date Key'].unique())
df['Is Delivered']= df['Invoice Date Key'].notna()
```

```python
zero_cust = df[df['Customer Key'] == 0]
print('Zero customer count:', len(zero_cust))
print('% of total rows: ', len(zero_cust)/ len(df) * 100)
```

**Data Validation**

```python
print('Product price check:', (round(df['Quantity'] * df['Unit Price'],2) == round(df['Total Excluding Tax'],2)).all())
print('Tax check:',(round(df['Total Excluding Tax'] + df['Tax Amount'],2) == round(df['Total Including Tax'],2)).all())
print('Item count check:',(df['Total Dry Items']+ df['Total Chiller Items'] == df['Quantity']).all())
```

```python
df['Expected Tax'] = round(df['Total Excluding Tax'] * df['Tax Rate'] / 100, 2)
tax_mismatch = df[round(df['Expected Tax'], 2) != round(df['Tax Amount'], 2)]
print('Which Tax Rate?', tax_mismatch['Tax Rate'].value_counts())
# 253 rows (~1%) show a fixed -0.01 deviation, all at 15% tax rate -> rounding rule difference, not an error
```

```python
# Negative profit investigation
neg_profit = df[df['Profit']<0]
print('Negative profit row count:', len(neg_profit))
print('% of total', len(neg_profit)/len(df) * 100)
```

```python
mask_count = neg_profit['Description'].value_counts()
print('Halloween mask total', mask_count[mask_count.index.str.contains('Halloween zombie mask')].sum())
print('Share of all negative rows:',mask_count[mask_count.index.str.contains('Halloween zombie mask')].sum()/len(neg_profit)*100)
```

```python
mask_all = df[df['Description'].str.contains('Halloween zombie mask', na=False)]
print("\nTotal sales of this product:", len(mask_all))
print("How many are negative profit:", (mask_all['Profit'] < 0).sum())
```

```python
mask_all = df[df['Description'].str.contains('Halloween zombie mask', na=False)].copy()
mask_all['Implied Unit Cost'] = mask_all['Unit Price'] - (mask_all['Profit'] / mask_all['Quantity'])

print("Average Unit Price:",mask_all['Unit Price'].mean())
print("Implied Unit Cost:", mask_all['Implied Unit Cost'].mean())
```

```python
df['Sold Below Cost'] = df['Profit'] < 0
```

```python
# Package column consistency
print("\nUnique Package values:", df['Package'].unique())
print("Any leading/trailing spaces?:", (df['Package'] != df['Package'].str.strip()).sum())
print("Case variations?:", (df['Package'] != df['Package'].str.title()).sum())
```
